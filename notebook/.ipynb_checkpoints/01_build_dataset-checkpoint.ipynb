First, we store index tickers in a Python dict keyed by `{Index Name: Ticker Symbol}`, then loop through each entry to:
1. **Fetch** — pull data from FRED
2. **Save** — write the raw response to `.csv`
3. **Process**

We convert the raw data to DataFrame before saving it purely for pipeline consistency, since every other source here is fetched or converted into DataFrame form before storage. Saving it as a Series instead would produce an identical CSV.

Here, the raw data is already clean, no further processing needed.

```python
fred = Fred(FRED_API_KEY)
fred_tickers = {
   'FFR_daily': 'DFF',
   'Palm_Oil': 'PPOILUSDM'
} 

for name, ticker in fred_tickers.items():
   series = fred.get_series(ticker, START, END)
   df = series.to_frame(name=name)
   df.index.name = 'Date'

   df.to_csv(f"{RAW_PATH}/{name}.csv")

   data_frames[name] = df
```

We pull the data from yfinance, save it, and then flatten yfinance's multi-index columns (the Price/Ticker header levels aren't needed here), retain only the daily closing price, and rename it to the index name

Since these operations are identical across all tickers, from here on we note only material deviations from this structure rather than repeating it in full.

```python
yahoo_tickers = {
   'KLCI': '^KLSE',
   'USDMYR': 'MYR=X',
   'Brent_Oil': 'BZ=F',
   'UST_10Y': '^TNX',
   'DXY': 'DX-Y.NYB',
   'VIX': '^VIX'
}

for name, ticker in yahoo_tickers.items():
   df = yf.download(ticker, START, END, progress=False)
   if df.empty:
      print(f"  FAILED - {ticker} returned no data, check the ticker code")
      continue

   df.to_csv(f'{RAW_PATH}/{name}.csv')
   
   print(f"{name} ({ticker}) - saved {len(df)} rows")

   if isinstance(df.columns, pd.MultiIndex):
      df.columns = df.columns.droplevel(1)

   df = df[['Close']].rename(columns={'Close': name})
   data_frames[name] = df
```

For each sector, we select 2-3 large, widely-held constituents (to reconstruct an equal-weighted index later on).

```python
sector_constituents = {
   'financials': ['1155.KL', '1023.KL', '1295.KL'],  # Maybank, CIMB, Public Bank
   'plantation': ['5285.KL', '1961.KL', '2445.KL'],  # Sime Darby Plantation, IOI Corp, KLK
   'reits':      ['5227.KL', '5176.KL', '5235SS.KL'],  # IGB REIT, Sunway REIT, KLCCP Stapled
   'technology': ['0166.KL', '0097.KL', '0128.KL'],  # Inari Amertron, ViTrox, Frontken
   'energy':     ['6033.KL', '5183.KL', '5681.KL'],  # PetGas, PetChem, Petronas Dagangan
   'industrial_products': ['8869.KL', '7113.KL'],    # Press Metal, Top Glove
}

for sector, tickers in sector_constituents.items():
   print(f"\n=== {sector.upper()} ===")
   closing_price = {}

   for t in tickers:
      df = yf.download(t, START, END, progress=False)
      
      if df.empty:
         print(f"  FAILED - {t} returned no data, check the ticker code")
         continue

      if isinstance(df.columns, pd.MultiIndex):
         df.columns = df.columns.droplevel(1)

      closing_price[t] = df['Close']
      print(f"  OK - {t} ({len(df)} rows)")

   sector_df = pd.DataFrame(closing_price)
   sector_df.index.name = 'Date'
   sector_df.to_csv(f'{RAW_PATH}/sectors/{sector}.csv')
   data_frames[sector] = sector_df
   print(f"  Saved {sector}.csv - {sector_df.shape[1]} of {len(tickers)} tickers resolved")
```

Now we fetch core CPI inflation data from DOSM, and replace the parquet's default numeric row index with `date`, since the numeric index carries no information.

Post-save, we narrow it down to overall division only (dropping the other divisions), and YoY inflation as our primary series - YoY strips out seasonal effects and matches how BNM and market commentary reference "inflation". Then adjust it back to p

One thing to note: core CPI data is only available from 2018 onwards, so YoY (needing a 12-month lookback) is only available from 2019 onwards - this affects CPI-related analysis specifically, not the full project's date range.

```python
df_cpi = pd.read_parquet('https://storage.dosm.gov.my/cpi/cpi_2d_core_inflation.parquet')

df_cpi['date'] = pd.to_datetime(df_cpi['date'])
df_cpi = df_cpi.set_index('date')

df_cpi.to_csv(f'{RAW_PATH}/CPI.csv')

df_cpi = df_cpi[df_cpi['division'] == 'overall']
df_cpi = df_cpi.loc[START:END]
df_cpi = df_cpi.drop(columns=['division', 'inflation_mom'])
df_cpi = df_cpi.dropna(how='any')
df_cpi = df_cpi.rename(columns={'inflation_yoy': 'cpi_inflation_yoy'})
```

As with core CPI, we set `date` as the index, then sort it. The year-by-year fetch doesn't guarantee chronological order, and an unsorted date index breaks range-based slicing later on.

Raw data is saved immediately after sorting, before any filtering. 

Post-save, we drop `year` (redundant once `date` is the index) and `change_in_opr` (we'll reconstruct this ourselves downstream), and rename `new_opr_level` to `OPR`.

```python
headers = {'Accept': 'application/vnd.BNM.API.v1+json'}

records = []
for year in range(START.year, END.year + 1):
   resp = requests.get(f'https://api.bnm.gov.my/public/opr/year/{year}', headers=headers)
   records.extend(resp.json()['data'])

df_opr = pd.DataFrame(records)

df_opr['date'] = pd.to_datetime(df_opr['date'])
df_opr = df_opr.set_index('date')
df_opr = df_opr.sort_index()

df_opr.to_csv(f'{RAW_PATH}/OPR.csv')

df_opr = df_opr.loc[START:END]

df_opr = df_opr.drop(columns=['year', 'change_in_opr'])
df_opr = df_opr.rename(columns={'new_opr_level': 'OPR'})
```
